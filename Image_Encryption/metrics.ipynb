```python
import time
import numpy as np
from pynq import Overlay, allocate  # <--- FIXED: Added 'allocate'

# 1. Setup
ol = Overlay("design_1.bit")
dma = ol.axi_dma_0

# Define sizes
N_OUTPUTS = 100000
input_buffer = allocate(shape=(1000,), dtype=np.int32)
output_buffer = allocate(shape=(N_OUTPUTS,), dtype=np.int32)

# 2. Measurement Loop
# Note: 32 bits per integer * N_OUTPUTS
DATA_SIZE_BITS = N_OUTPUTS * 32  
start_time = time.time()

# Trigger the transfer
dma.sendchannel.transfer(input_buffer)
dma.recvchannel.transfer(output_buffer)

# Wait for completion
dma.sendchannel.wait()
dma.recvchannel.wait()

end_time = time.time()

# 3. Calculation
duration = end_time - start_time
throughput_bps = DATA_SIZE_BITS / duration
throughput_gbps = throughput_bps / 1e9

print(f"Data Generated: {DATA_SIZE_BITS / 8 / 1024:.2f} KB")
print(f"Time Taken:     {duration:.4f} s")
print(f"Throughput:     {throughput_gbps:.4f} Gbps")

# 4. Cleanup (Good practice)
del input_buffer
del output_buffer
```
